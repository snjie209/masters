## Simulate sinusoidal data

According to this [paper on cosinor-based rhythmometry](https://tbiomed.biomedcentral.com/articles/10.1186/1742-4682-11-16):

$$Y(t) = M + A \text{cos}\left ( 2\pi t / \tau + \phi \right ) + e(t),$$

where:

+ $Y(t)$ is the number of gene transcripts, 

+ $M$ is the mesor, or the mean value of the sine wave,

+ $A$ is the amplitude, 

+ $\tau$ is the period, 

+ $\phi$ is the acrophase which is the time to the peak of the sine wave.

Specifically for the [Tempo paper](https://www.nature.com/articles/s41467-022-34185-w#Sec8), we adhere to the following formulas:

For $i$ representing the cell index, $j$ the gene index, the UMI count $X_{ij}$ follows a negative binomial distribution:

$$X_{ij} \sim \text{NB}\left ( L_{i}\lambda_{ij}, \delta_{ij} \right ),$$

where:

1. $\mathbb{E}\left [ X_{ij} \right ] = L_{i}\lambda_{ij}$, 

2. $\text{Var}\left [ X_{ij} \right ] = \mathbb{E}\left [ X_{ij} \right ] + \delta_{ij} \left ( \mathbb{E}\left [ X_{ij} \right ] \right )^{2}$,

3. $\text{log}(\lambda_{ij}) = \mu_{j} + Q_{j}A_{j}\text{cos}\left ( \theta_{i} - \phi_{j} \right )$, and

4. $\delta_{ij} = \text{g}_{\zeta}(\lambda_{ij})$

The terms are

+ $L_{i}$ is the size of cell library of the cell $i$, 

+ $\text{log} \: \lambda_{ij}$ is the log-proportion gene $j$'s transcripts in cell $i$, 

+ $\mu_{j}$ is the mesor of gene $j$, 

+ $A_{j}$ is the amplitude of gene $j$, 

+ $\phi_{j}$ is the acrophase of gene $j$, 

+ $\theta_{j}$ is the **latent cell phase*** we are trying to estimate for cell $i$, 

+ $Q_{j}$ describes whether the gene has non-zero amplitude, 

+ $\text{g}_{\zeta}$ is a deterministic polynomial function relating $\lambda_{ij}$ with the dispersion parameter of $X_ij$ UMI counts, parameterized by $\zeta$ which is across all genes and cells. 


To sample, we will get the outputs of the Tempo model, which are the posterior gene parameters, to sample the sinusoid parameters which will be used as inputs to the sine function, and with a vector of timepoints, create sinusoids per gene.
