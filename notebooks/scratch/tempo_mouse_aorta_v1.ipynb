## Simulate sinusoidal data

### Simulate data using Tempo

According to this [paper on cosinor-based rhythmometry](https://tbiomed.biomedcentral.com/articles/10.1186/1742-4682-11-16):

$$Y(t) = M + A \text{cos}\left ( 2\pi t / \tau + \phi \right ) + e(t),$$

where:

+ $Y(t)$ is the number of gene transcripts, 

+ $M$ is the mesor, or the mean value of the sine wave,

+ $A$ is the amplitude, 

+ $\tau$ is the period, 

+ $\phi$ is the acrophase which is the time to the peak of the sine wave.

Specifically for the [Tempo paper](https://www.nature.com/articles/s41467-022-34185-w#Sec8), we adhere to the following formulas:

For $i$ representing the cell index, $j$ the gene index, the UMI count $X_{ij}$ follows a negative binomial distribution:

$$X_{ij} \sim \text{NB}\left ( L_{i}\lambda_{ij}, \delta_{ij} \right ),$$

where:

1. $\mathbb{E}\left [ X_{ij} \right ] = L_{i}\lambda_{ij}$, 

2. $\text{Var}\left [ X_{ij} \right ] = \mathbb{E}\left [ X_{ij} \right ] + \delta_{ij} \left ( \mathbb{E}\left [ X_{ij} \right ] \right )^{2}$,

3. $\text{log}(\lambda_{ij}) = \mu_{j} + Q_{j}A_{j}\text{cos}\left ( \theta_{i} - \phi_{j} \right )$, and

4. $\delta_{ij} = \text{g}_{\zeta}(\lambda_{ij})$

The terms are

+ $L_{i}$ is the size of cell library of the cell $i$, 

+ $\text{log} \: \lambda_{ij}$ is the log-proportion gene $j$'s transcripts in cell $i$, 

+ $\mu_{j}$ is the mesor of gene $j$, 

+ $A_{j}$ is the amplitude of gene $j$, 

+ $\phi_{j}$ is the acrophase of gene $j$, 

+ $\theta_{j}$ is the **latent cell phase** we are trying to estimate for cell $i$, 

+ $Q_{j}$ describes whether the gene has non-zero amplitude, 

+ $\text{g}_{\zeta}$ is a deterministic polynomial function relating $\lambda_{ij}$ with the dispersion parameter of $X_ij$ UMI counts, parameterized by $\zeta$ which is across all genes and cells. 


To sample, we will get the outputs of the Tempo model, which are the posterior gene parameters, to sample the sinusoid parameters which will be used as inputs to the sine function, and with a vector of timepoints, create sinusoids per gene.

```python
# Configs
N_SAMPLES = 1000 # of unique timepoints sampled
MIN_T = 0 # minimum of the time range
MAX_T = 100 # maximum of the time range
TAU = 24 # period

# Get max and min amplitudes (since thisis of beta which has support [0, 1] we set it to this for now)
A_MIN = 0 # min amplitude we want to set across waveforms
A_MAX = 1 # max amplitude we want to set across waveforms
```

```python
def sample_timepoints(n_samples = 100, 
                      min_t = 0, 
                      max_t = 10, 
                      size = 1):
    
    return np.linspace(min_t, max_t, n_samples)

def sample_gene_params(gene_dist_row: pd.Series, 
                       A_max: int, A_min: int, size: int = 1):
    # # Sample phase from VonMises distribution:
    # phase_i = np.random.vonmises(mu = gene_dist_row['mu_loc'], 
    #                           kappa = gene_dist_row['mu_scale'], size = 1)
    
    # Sample mesor
    mu_j = np.random.normal(loc = gene_dist_row['mu_loc'],
                            scale = gene_dist_row['mu_scale'],
                            size = size)
    
    # Sample amplitude
    A_j = (np.random.beta(a = gene_dist_row['A_alpha'], 
                          b = gene_dist_row['A_beta'],
                          size = size)
          ) * (A_max - A_min) + A_min
    
    # Sample acrophase from a Von Mises if prior info known. 
    # If no prior info, we sample from a hyperspherical uniform prior
    phi_j = np.random.vonmises(mu = gene_dist_row['phi_loc'],
                               kappa = gene_dist_row['phi_scale'], 
                               size = size)
    
    # if size == 1, then output is Array([scalar]) so we need to turn it into a scalar in this case
    if size == 1: 
        mu_j, A_j, phi_j = mu_j[0], A_j[0], phi_j[0]
    
    return mu_j, A_j, phi_j


def generate_gene_sine(mu_j, A_j, phi_j, t, tau = 24):
    """
    Generates a sample cell gene expression data (t for x-axis timesteps, y for the values) 
    for given waveform parameters for gene j.
    """

    y_j = mu_j + A_j * np.cos(2 * np.pi * t / tau + phi_j)

    return y_j

def generate_sine(gene_params_df, tau = 24, n_samples = 1000, min_t = 0, max_t = 20):
    """
    Generates waveforms for each gene across various genes.
    """

    # Generate timepoints which will form the X-axis for each waveform
    t_vec = sample_timepoints(n_samples = n_samples, min_t = min_t, max_t = max_t)

    # Place to store waveforms for each gene
    waveforms = dict()

    for gene_index in gene_params_df.index:
        gene_params = gene_params_df.loc[gene_index, :]
        mu = gene_params["mu"]
        A = gene_params["A"]
        phi = gene_params["phi"]

        waveforms[gene_index] = generate_gene_sine(mu_j = mu, 
                                                   A_j = A, 
                                                   phi_j = phi, 
                                                   tau = tau,
                                                   t = t_vec)
        
    # Output as dataframe where columns = each gene, rows = each timepoint, values = gene expression counts
    time_df = pd.DataFrame(dict(time = t_vec))
    waves_df = pd.DataFrame(waveforms)
    waves_df = pd.concat([time_df, waves_df], axis = 1)
    return waves_df

def simulate_tempo_data(gene_dist_df, 
                        n_samples, 
                        min_t, 
                        max_t, 
                        A_min, 
                        A_max, 
                        tau = 24, size = 1):
    mu_vec, A_vec, phi_vec = list(), list(), list()

    for gene_idx in gene_dist_df.index:
        gene_dist_row = gene_dist_df.loc[gene_idx, :]
        
        # Sample sinusoid params
        mu_j, A_j, phi_j = sample_gene_params(gene_dist_row = gene_dist_row, 
                                              A_min = A_min, A_max = A_max, size = size)

        # Store in a vector for each gene iteration
        mu_vec.append(mu_j)
        A_vec.append(A_j)
        phi_vec.append(phi_j)

    # Store all gene params in a dataframe
    gene_params_df = pd.DataFrame({"idx" : gene_dist_df.index,
                                    "mu" : mu_vec, 
                                    "A" : A_vec, 
                                    "phi" : phi_vec})
    gene_params_df = gene_params_df.set_index("idx")

    # Generate sinusoidal realizations
    waves_df = generate_sine(gene_params_df = gene_params_df, 
                             n_samples = n_samples, 
                             min_t = min_t,
                             max_t = max_t,
                             tau = tau)
    
    return waves_df

```

## Gaussian Process (MOGPTK) Modeling

+ [MOGPTK Paper](https://arxiv.org/abs/2002.03471)
+ [MOGPTK Github](https://github.com/GAMES-UChile/mogptk)
+ [Understanding Multi-Output Gaussian Process Methods (ICM, SLFM, LCM)](https://gpss.cc/gpss17/slides/multipleOutputGPs.pdf)
+ [MOSM Paper One Type of Kernel](https://arxiv.org/abs/1709.01298)


By construction, a Gaussian process is a stochastic process that is sampled from a multivariate normal distribution, defined by a mean function $\mu(x)$ and covariance function $k(x, x')$ such that for any finite set of inputs (perhaps time points) $\left \{ x_{i} \right \}_{i = 1, ..., n}$ we can apply mean and covariance functions to it, constructing an $N \times 1$ vector for the mean function output, and $N \times N$ matrix for the covariance matrix which is a result of the covariance function applied $N^{2}$ times. Training a Gaussian process model would then mean learning the parameters of this covariance kernel function, and inference would mean, as is done with a generative model, sampling from the learned mean and covariance function. 

Above we discussed univariate Gaussian process, but the multivariate case would consider $K$ stochastic processes (i.e., $K$ time series) that are assumed to be coupled, and prediction/inference could mean forecasting future timepoints for all $K$ timeseries. This would, in lieu, be one way to determine coupling, albeit indirectly.

## Neural Controlled Differential Equations (NCDE)

Similar to MOGPs, neural ordinary differential equations can be treated as a forecasting model where we learn parameters for a model using historical data from $K$ time series, and forecast future points across all $K$ time series. 

The difference here though is how the time series forecast is exactly done. With a MOGP, we learn a kernel function that enables us to sample, from a multi-variate Gaussian, essentially each time series for each draw. For a NCDE, we treat the set of time series as a dynamical system that is governed by a (set of) differential equation(s), and we use NCDE as a mechanism to evaluate those diffeq(s) numerically to find the solutions ($y$ values) that can be the predictions.

### NCDE Functions

```python
# First, define the function which will specify the vector field of
# the differential equation.
class Func(eqx.Module):
    mlp: eqx.nn.MLP
    data_size: int
    hidden_size: int

    def __init__(self, data_size, hidden_size, width_size, depth, *, key, **kwargs):
        super().__init__(**kwargs)
        self.data_size = data_size
        self.hidden_size = hidden_size
        self.mlp = eqx.nn.MLP(
            hidden_size,
            hidden_size * data_size,
            width_size,
            depth,
            activation=jnn.tanh,
            # Note the use of a tanh final activation function. This is important to
            # stop the model blowing up. (Just like how GRUs and LSTMs constrain the
            # rate of change of their hidden states.)
            final_activation=jnn.tanh,
            key=key,
        )

    def __call__(self, t, y, args):
        return self.mlp(y).reshape(self.hidden_size, self.data_size)
    
# The above Func will be solved as part of the NeuralCDE algo
# Below. Note we do have an MLP for the initial condition
# As well as the future iterations.

class NeuralCDE(eqx.Module):
    initial: eqx.nn.MLP
    func: Func
    linear: eqx.nn.Linear

    def __init__(self, data_size, hidden_size, width_size, output_size, depth, *, key, **kwargs):
        super().__init__(**kwargs)
        ikey, fkey, lkey = jr.split(key, 3)
        self.initial = eqx.nn.MLP(in_size = data_size, 
                                  out_size = hidden_size, 
                                  width_size = width_size, 
                                  depth = depth, 
                                  key = ikey)
        self.func = Func(data_size, hidden_size, width_size, depth, key=fkey)
        self.linear = eqx.nn.Linear(hidden_size, output_size, key=lkey)

    def __call__(self, ts, coeffs, evolving_out=False):
        # Each sample of data consists of some timestamps `ts`, and some `coeffs`
        # parameterising a control path. These are used to produce a continuous-time
        # input path `control`.
        control = diffrax.CubicInterpolation(ts, coeffs)
        print(f"Finished control with {control}")
        term = diffrax.ControlTerm(self.func, control).to_ode()
        solver = diffrax.Tsit5()
        dt0 = None
        y0 = self.initial(control.evaluate(ts[0]))

        print(f"ts[0] shape: {ts[0].shape}")
        # print(f"y0 shape: {y0.shape}")
        if evolving_out:
            saveat = diffrax.SaveAt(ts=ts)
        else:
            saveat = diffrax.SaveAt(t1=True)
        solution = diffrax.diffeqsolve(
            term,
            solver,
            ts[0],
            ts[-1],
            dt0,
            y0,
            stepsize_controller=diffrax.PIDController(rtol=1e-3, atol=1e-6),
            saveat=saveat,
        )
        
        # return solution.ys, y0
        if evolving_out:
            prediction = jax.vmap(lambda y: jnn.sigmoid(self.linear(y))[0])(solution.ys)
        else:
            # print(f"solution.ys has shape {solution.ys.shape}")
            prediction = jnn.sigmoid(self.linear(solution.ys[-1])) # (prediction, )
        return prediction

# Data Processing Functions
def dataloader(arrays, batch_size, *, key):
    dataset_size = arrays[0].shape[0]
    assert all(array.shape[0] == dataset_size for array in arrays)
    indices = jnp.arange(dataset_size)
    while True:
        perm = jr.permutation(key, indices)
        (key,) = jr.split(key, 1)
        start = 0
        end = batch_size
        while end < dataset_size:
            batch_perm = perm[start:end]
            yield tuple(array[batch_perm] for array in arrays)
            start = end
            end = start + batch_size

def window_stack(df, width=3, stepsize=1):
    """
    Takes in waves_df (T, G) dimension (T = # timepoints, G = # genes)
    and slides a window of length width (W), sliding over on stepsize at a time,
    to return two datasets:

    1. Jax array of feature vectors, size (T - W, G, W)
    2. Jax array of target vectors, size (T - W, G, 1)
    """

    dat = df.drop(labels = 'time', axis = 1) # remove time column
    dat = dat.T # transpose from [time, genes] to [genes, time] 
    dat = dat.values # only gets values, which will make us lose gene index
    max_T = dat.shape[1] # number of timepoints
    range_T = df.time.to_list() # store all timepoints in a list

    # Create windowed Jax array for features (technically autoregressors)
    feat_window_list = [dat[:, k: (k + width)] for k in range(0, max_T - width, stepsize)] 
    feat_window_jnp = jnp.array(feat_window_list) # turns this into a Jax array of size [T - W, G, W]
    feat_window_jnp = jnp.moveaxis(feat_window_jnp, -1, -2) # transposes last two axis so we get [T - W, W, G]

    # Prep timepoints to be stacked with the features, as this is a requirement for the diffrax training method
    time_window_jnp = jnp.array([range_T[k:k + width] for k in range(0, max_T - width, stepsize)]) # size [T - W, W]
    time_windowed = jnp.expand_dims(time_window_jnp, -1) # to [T - W, W, 1]

    # Concatenate (T - W, W, G) tensor of features with (T - W, W, 1) tensor of time to get a (T - W, W, G + 1) tensor of features + time
    x_windowed = jnp.concatenate([time_windowed, feat_window_jnp], axis = -1)

    # Create windowed Jax array for target (= response) variables
    target_window_list = [dat[:, k + width] for k in range(0, max_T - width, stepsize)] 
    target_window_jnp = jnp.array(target_window_list) # turns into a Jax array of size [T - W, G]
    y_windowed = jnp.expand_dims(target_window_jnp, -2) # makes it same dimensionality as feat_window_jnp, i.e., [T - W, G] to [T - W, 1, G]

    return x_windowed, y_windowed, time_windowed

# Modeling Functions
def train_test_split(X, y, t, pct_hist):
    max_T = X.shape[0]
    hist_cutoff = int(max_T * pct_hist)

    X_hist, y_hist, t_hist = X[:hist_cutoff, :, :], y[:hist_cutoff, :, :], t[:hist_cutoff, :, :]
    X_future, y_future, t_future = X[hist_cutoff:, :, :], y[hist_cutoff:, :, :], t[hist_cutoff:, :, :] 

    return X_hist, y_hist, t_hist, X_future, y_future, t_future


def cubic_spline_interpolation(X_stack, t_stack):
    # Do cubic spline interpolation to get X
    if t_stack.ndim == 3:
        t_stack = t_stack[:, :, 0]

    coeffs = jax.vmap(diffrax.backward_hermite_coefficients)(t_stack, X_stack)

    return coeffs

```

```python
# Training loop like normal.

def compute_mse(y_true, y_pred, epsilon=1e-15):
    """
    MSE Loss for forecasting with real numbers
    """
    if y_true.ndim == 3:
        y_true = y_true[:, 0, :]

    # Ensure predictions are clipped to avoid log(0)
    # y_pred = jnp.clip(y_pred, epsilon, 1 - epsilon)

    # Calculate the categorical cross-entropy
    loss = jnp.mean(optax.l2_loss(y_pred, y_true))
    # loss = -jnp.mean((y_true - y_pred)**2, axis=1)

    # Return the mean loss over all samples
    return jnp.mean(loss)

def ncde_predict(model, ti, coeff_i):
    return jax.vmap(model)(ti, coeff_i)

@eqx.filter_jit
def loss(model, ti, y_true, coeff_i):
    pred = ncde_predict(model, ti, coeff_i)

    # MSE
    mse_val = compute_mse(y_true = y_true, y_pred = pred)

    # mse_val = jnp.mean(mse_val)

    return mse_val, mse_val

# has_aux true will return two values: output to be differentiated and its gradient
grad_loss = eqx.filter_value_and_grad(loss, has_aux=True) 

@eqx.filter_jit
def make_step(model, data_i, opt_state):
    ti, label_i, *coeff_i = data_i

    (mse_val, _), grads = grad_loss(model, ti, label_i, coeff_i)
    updates, opt_state = optim.update(grads, opt_state)
    model = eqx.apply_updates(model, updates)
    return mse_val, model, opt_state

# Optional section
optim = optax.adam(lr)
opt_state = optim.init(eqx.filter(model_sim, eqx.is_inexact_array))
for step, data_i in zip(
    range(steps), dataloader((t_hist[:, :, 0], y_hist) + coeffs_hist, BATCH_SIZE, key=loader_key)
):
    start = time.time()
    mse_val, model_sim, opt_state = make_step(model_sim, data_i, opt_state)
    end = time.time()
    print(
        f"Step: {step}, Loss: {mse_val}, Accuracy: {mse_val}, Computation time: "
        f"{end - start}"
    )

# ts, coeffs, labels, _ = get_data(dataset_size, add_noise, key=test_data_key)
(mse_val, _) = loss(model = model_sim,
                    ti = t_hist[:, :, 0],
                    y_true = y_hist,
                    coeff_i = coeffs_hist)
print(f"Test loss: {mse_val}, Test Accuracy: {mse_val}")

# Plot results
sample_ts = t_hist[:, :, 0][-1]
sample_coeffs = tuple(c[-1] for c in coeffs_hist)
pred = model_sim(sample_ts, sample_coeffs, evolving_out=True)
interp = diffrax.CubicInterpolation(sample_ts, sample_coeffs)
values = jax.vmap(interp.evaluate)(sample_ts)
fig = plt.figure(figsize=(16, 8))
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2, projection="3d")
ax1.plot(sample_ts, values[:, 1], c="dodgerblue")
ax1.plot(sample_ts, values[:, 2], c="dodgerblue", label="Data")
ax1.plot(sample_ts, pred, c="crimson", label="Classification")
ax1.set_xticks([])
ax1.set_yticks([])
ax1.set_xlabel("t")
ax1.legend()
ax2.plot(values[:, 1], values[:, 2], c="dodgerblue", label="Data")
ax2.plot(values[:, 1], values[:, 2], pred, c="crimson", label="Classification")
ax2.set_xticks([])
ax2.set_yticks([])
ax2.set_zticks([])
ax2.set_xlabel("x")
ax2.set_ylabel("y")
ax2.set_zlabel("Classification")
plt.tight_layout()
plt.savefig("neural_cde.png")
plt.show()
```

```python
def batch_generate_waveforms(gene_dist_df, n_timepoints, min_t, max_t, n_waves = 1, tau = 24, A_min = 0, A_max = 1):
    """
    Get a matrix of size (n_waves, n_timepoints, n_genes) where n_genes is number of genes in the gene_dist_df index.
    """
    # Place to store waveforms for each gene
    waveforms = []

    # number of timepoints (shape: (n_samples,))
    t = sample_timepoints(n_samples = n_timepoints, min_t = min_t, max_t = max_t)

    for gene in gene_dist_df.index:
        # One pandas Series row of the DataFrame with the 
        gene_dist_row = gene_dist_df.loc[gene, :]

        # each parameter (mu_j, A_j, phi_j) is of shape (n_waves, )
        mu_j, A_j, phi_j = sample_gene_params(gene_dist_row = gene_dist_row, A_min = A_min, A_max = A_max, size = n_waves)
        
        # will repeat values of t_vec (sized (n_timepoints,)) to be (n_timepoints, n_waves)
        t_mat = np.broadcast_to(t, (n_waves, n_timepoints)).T

        # Get a numpy 2d matrix of waveforms where each row is a waveform with 'n_timepoints' timepoints, and there are 'n_waves' many rows.
        waveform_mat = mu_j + A_j * np.cos(2 * np.pi * t_mat / tau + phi_j)

        waveforms.append(waveform_mat)

    # If n_genes is number of genes, or length of index in gene_dist_df, 
    # below would be of shape (n_genes, n_timepoints, n_waves)
    waveforms = np.array(waveforms)
    # Transpose (n_genes, n_timepoints, n_waves) to (n_waves, n_timepoints, n_genes)
    waveforms = np.transpose(waveforms, (2, 1, 0))

    return waveforms
```
